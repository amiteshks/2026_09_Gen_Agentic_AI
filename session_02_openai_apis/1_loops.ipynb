```python
#Create a While loop and break the loop if the entered number is an odd number
while True:
    number = int(input("Please enter a number: "))
    if (number % 2) != 0:
        print(f"{number} is an odd number. Exiting the loop.")
        break
    else:
        print(f"{number} is an even number. Please try again.")
```
