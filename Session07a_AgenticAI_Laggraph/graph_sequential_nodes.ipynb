### Build a simple sequential graph using Langgraph

```python
from langgraph.graph import StateGraph, START, END
```

```python
def fetch_claim_details_agent_function(dict):
    # Simulate some processing
    # Do something useful here
    print("Fetching claim details ....")
    print("Claim Details: Claim ID: 12345, Claimant: John Doe, Amount: $5000, Status: Pending.")
    return {}

def summarize_claim_details_agent_function(dict):
    # Simulate some processing
    # Do something useful here
    print("Summarizing claim details ....")
    print("Claim Summary: This is a summary of the claim details.")
    print("Claim Summary: The claim is valid and requires further processing.")
    return {}

builder = StateGraph(dict)


builder.add_node("Fetch_Claim_Details", fetch_claim_details_agent_function)
builder.add_node("Summarize_Claim_Details", summarize_claim_details_agent_function)

builder.add_edge(START, "Fetch_Claim_Details")
builder.add_edge("Fetch_Claim_Details", "Summarize_Claim_Details")

builder.add_edge("Summarize_Claim_Details", END)

graph = builder.compile()
```

```python

# --- Display as ASCII ---
print(graph.get_graph().draw_ascii())
```

```python
# Invoke the graph with an empty initial state
initial_state = {}
graph.invoke(initial_state)
```
