```python
import os
from dotenv import load_dotenv

load_dotenv(override=True, dotenv_path="../.env.local")
my_api_key = os.getenv("OPENAI_API_KEY")
print(my_api_key[-5:]) 
```

```python
from openai import OpenAI
client = OpenAI(api_key=my_api_key)
```

```python
prompt = "What's an ETF?"#input("Ask me a question about investments: ")

response = client.chat.completions.create( #Chat Completion Create API Call
    model="gpt-4.1-nano", 
    messages=[
        {"role": "system", "content": "User may ask questions about investments. You are a Financial expert."},
        {"role": "user", "content": prompt}
    ]
)
print (response.choices[0].message.content)
```
