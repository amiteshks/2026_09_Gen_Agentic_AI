```python
import os
import asyncio

from dotenv import load_dotenv
from openai import OpenAI

from mcp import ClientSession
from mcp.client.streamable_http import streamable_http_client

# Remote MCP Server
SERVER = "https://mcp.deepwiki.com/mcp"

# Your public GitHub repository
REPO = "amiteshks/2026_09_Gen_Agentic_AI"

print("MCP Server:", SERVER)
print("GitHub Repository:", REPO)
```

```python
load_dotenv(
    override=True,
    dotenv_path="../../.env.local"
)

my_api_key = os.getenv("OPENAI_API_KEY")

client = OpenAI(api_key=my_api_key)

print("OpenAI client initialized")
```

```python
response = client.responses.create(
    model="gpt-6-luna",
    tools=[
        {
            "type": "mcp",
            "server_label": "deepwiki",
            "server_url": "https://mcp.deepwiki.com/mcp",
            "require_approval": "never"
        }
    ],
    input=(
        "Explain the architecture of the GitHub repository "
        "amiteshks/2026_09_Gen_Agentic_AI. "
        "Identify its major components and explain how they interact."
    )
)

print(response.output_text)
```

```python
for item in response.output:

    if item.type == "mcp_list_tools":
        print("\nDISCOVERED MCP TOOLS")

        for tool in item.tools:
            print(tool.name)
            print(tool.description)

    elif item.type == "mcp_call":
        print("\nMCP TOOL CALLED")
        print("Tool:", item.name)
        print("Arguments:", item.arguments)
        print("Result:", item.output)
```
